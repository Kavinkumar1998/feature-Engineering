## import necsary lib"

```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import  accuracy_score,classification_report,confusion_matrix,recall_score,roc_auc_score,precision_score,roc_curve
from sklearn.ensemble import GradientBoostingClassifier 
from sklearn.feature_selection import SelectFromModel,RFE
from sklearn.ensemble import RandomForestClassifier
```

## 2.Import the dataset

```python
breast_cancer_data=load_breast_cancer()
```

```python
breast_cancer_dataframe = pd.DataFrame(data=breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_dataframe["Target"]=breast_cancer_data.target
breast_cancer_dataframe
```

## 3.data understanding

```python
breast_cancer_dataframe.describe()
```

```python
breast_cancer_dataframe.dtypes
```

```python
breast_cancer_dataframe.isna().sum()
```

## 4.Data Preparation

```python
X=breast_cancer_dataframe.drop(labels="Target",axis= 1)
X
```

```python
y=breast_cancer_dataframe["Target"]
y
```

```python
X_train, X_test, y_train, y_test =train_test_split(X,y,test_size=0.20,train_size=0.80,random_state=12,shuffle=True,stratify=y)
```

## model building testing

```python
rf=RandomForestClassifier(max_depth=4)
```

```python
rf.fit(X_train,y_train)
```

```python
y_pred_train = rf.predict(X_train)
y_pred_test = rf.predict(X_test)
```

```python
print(classification_report(y_train,y_pred_train))
print("Accuracy Score for tarining data :" ,accuracy_score(y_train,y_pred_train))
print("Confusion Matrix fir training Data: \n", confusion_matrix(y_train,y_pred_train))
print("Precision Score : ",precision_score(y_train,y_pred_train))
print("Recall Score :", recall_score(y_train,y_pred_train))
```

```python
print(classification_report(y_test,y_pred_test))
print("Accuracy Score for tarining data :" ,accuracy_score(y_test,y_pred_test))
print("Confusion Matrix fir training Data: \n", confusion_matrix(y_test,y_pred_test))
print("Precision Score : ",precision_score(y_test,y_pred_test))
print("Recall Score :", recall_score(y_test,y_pred_test))
```

```python
rf.feature_importances_
```

```python
sfm=SelectFromModel( estimator=RandomForestClassifier(),threshold=None,prefit=False,norm_order=1,max_features=12)
```

```python
sfm.fit(X_train,y_train)
```

```python
sfm.get_support()
```

```python
breast_cancer_data.feature_names[sfm.get_support()]
```

```python
X_train_sfm=sfm.transform(X_train)
X_test_sfm=sfm.transform(X_test)
```

```python
rf.fit(X_train_sfm,y_train)
y_pred_train_sfm = rf.predict(X_train_sfm)
y_pred_test_sfm = rf.predict(X_test_sfm)
```

```python
print(classification_report(y_train,y_pred_train_sfm))
print("Accuracy Score for tarining data :" ,accuracy_score(y_train,y_pred_train_sfm))
print("Confusion Matrix fir training Data: \n", confusion_matrix(y_train,y_pred_train_sfm))
print("Precision Score : ",precision_score(y_train,y_pred_train_sfm))
print("Recall Score :", recall_score(y_train,y_pred_train_sfm))
```

```python
print(classification_report(y_test,y_pred_test_sfm))
print("Accuracy Score for tarining data :" ,accuracy_score(y_test,y_pred_test_sfm))
print("Confusion Matrix fir training Data: \n", confusion_matrix(y_test,y_pred_test_sfm))
print("Precision Score : ",precision_score(y_test,y_pred_test_sfm))
print("Recall Score :", recall_score(y_test,y_pred_test_sfm))
```
